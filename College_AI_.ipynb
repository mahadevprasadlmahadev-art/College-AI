```python
# 2. Direct Short Answer Engine (Single Answer, Zero Thinking)
def get_direct_short_answer(message: str) -> str:
    q = message.lower().strip()
    if any(k in q for k in ["fee", "cost", "tuition", "annual", "charge", "structure", "payment", "how much"]):
        return (
            "💰 **Annual Fee (2024–2028):** **₹1,60,000 per year**\n"
            "• **Tuition Fee:** ₹1,15,956\n"
            "• **Other Fees:** ₹44,044\n"
            "*Note: Fee is common for all engineering departments.*"
        )
    for day in ["monday", "tuesday", "wednesday", "thursday", "friday"]:
        if day in q:
            day_cap = day.capitalize()
            slots = COLLEGE_DATA["timetable"][day_cap]
            lines = [f"• `{t}` : **{s}**" for t, s in slots]
            return f"📅 **{day_cap} Timetable (5th Sem AI & DS):**\n" + "\n".join(lines)
    if "network" in q or "bcs302" in q:
        return "👨‍🏫 **Faculty:** **Prof. Harshitha BS** (`HBS`) teaches **Computer Networks** (BCS302)."
    if "cloud" in q or "bas305" in q:
        return "👨‍🏫 **Faculty:** **Dr. Swarnalatha K** (`SK`) teaches **Cloud Computing** (BAS305)."
    if "software" in q or "sepm" in q or "bas301" in q:
        return "👨‍🏫 **Faculty:** **Prof. Madhuri B** (`MB`) teaches **Software Engineering** (BAS301)."
    if any(k in q for k in ["faculty", "teacher", "professor", "who teaches"]):
        return (
            "👨‍🏫 **MITT Faculty Directory (5th Sem AI & DS):**\n"
            "• **Prof. Madhuri B** (MB) — Software Engineering\n"
            "• **Prof. Harshitha BS** (HBS) — Computer Networks\n"
            "• **Prof. Mamatha D** (MD) — Theory of Computation & DV Lab\n"
            "• **Dr. Swarnalatha K** (SK) — Cloud Computing\n"
            "• **Dr. GS Madhan Kumar** (GSM) — Mini Project & Research Methodology\n"
            "• **Prof. Mahadev Prasad N** (MPN) — Environmental Studies"
        )
    if any(k in q for k in ["subject", "code", "vtu", "course"]):
        return (
            "📚 **5th Sem Subjects & VTU Codes:**\n"
            "• Software Engineering: `BAS301`\n"
            "• Computer Networks: `BCS302`\n"
            "• Theory of Computation: `BCS303`\n"
            "• Data Visualization Lab: `BAIL304`\n"
            "• Cloud Computing: `BAS305`\n"
            "• Mini Project: `BAS306`\n"
            "• Research Methodology: `BCS307`\n"
            "• Environmental Studies: `BIOC308`\n"
            "• Yoga: `BYOK309`"
        )
    if any(k in q for k in ["start", "commence", "working day", "semester", "dates"]):
        return "🎓 **5th Sem Dates:** Classes Start: **08/09/2026** | Last Working Day: **30/12/2026**"
    return "👋 Ask me about 💰 Fees, 📅 Timetables, 👨‍🏫 Faculty, or 📚 VTU Codes!"

def stream_ai_response(message: str, history=None):
    cleaned_msg = message.strip()
    if not cleaned_msg:
        return
    ans = get_direct_short_answer(cleaned_msg)
    words = ans.split(" ")
    for i, w in enumerate(words):
        yield w + (" " if i < len(words) - 1 else "")
        time.sleep(0.012)
```

```python
# 3. Test Response Speed & Content
t0 = time.time()
print("Query: What is the annual fee for 2026-27?\n")
ans = ""
for chunk in stream_ai_response("What is the annual fee for 2026-27?"):
    ans += chunk
print(ans)
print(f"\n⚡ Total Latency: {time.time()-t0:.3f}s (Instant, Single Direct Answer)")
```

```python
# 4. Launch Attractive 60-30-10 Gradio Interface
try:
    import gradio as gr
    HAS_GRADIO = True
except ImportError:
    HAS_GRADIO = False

if not HAS_GRADIO:
    print("Install Gradio: pip install gradio")
else:
    def chat_fn(message, history):
        if not message or not message.strip():
            yield "", history
            return
        user_msg = message.strip()
        history = history or []
        updated_history = list(history) + [
            {"role": "user", "content": user_msg},
            {"role": "assistant", "content": ""}
        ]
        yield "", updated_history
        bot_text = ""
        for chunk in stream_ai_response(user_msg, history):
            bot_text += chunk
            updated_history[-1]["content"] = bot_text
            yield "", updated_history

    with gr.Blocks(title="MITT College AI Assistant") as demo:
        gr.Markdown("# ⚡ MITT College AI Assistant\n**Instant Direct Answers • Zero Thinking Delay**")
        with gr.Row():
            b1 = gr.Button("💰 Annual Fees")
            b2 = gr.Button("📅 Monday Schedule")
            b3 = gr.Button("👨‍🏫 Faculty Directory")
            b4 = gr.Button("📚 VTU Codes")
        chatbot = gr.Chatbot(height=480, type="messages")
        with gr.Row():
            msg = gr.Textbox(placeholder="Ask anything about MITT College...", scale=8, show_label=False)
            send = gr.Button("➤ Send", variant="primary", scale=2)
        clear = gr.Button("🗑️ Clear Chat", size="sm")
        send.click(chat_fn, [msg, chatbot], [msg, chatbot])
        msg.submit(chat_fn, [msg, chatbot], [msg, chatbot])
        clear.click(lambda: [], outputs=chatbot)
        b1.click(lambda: "What is the annual fee for 2026-27?", outputs=msg).then(chat_fn, [msg, chatbot], [msg, chatbot])
        b2.click(lambda: "What is my Monday timetable?", outputs=msg).then(chat_fn, [msg, chatbot], [msg, chatbot])
        b3.click(lambda: "List all faculty members.", outputs=msg).then(chat_fn, [msg, chatbot], [msg, chatbot])
        b4.click(lambda: "Give me the list of subjects and VTU codes.", outputs=msg).then(chat_fn, [msg, chatbot], [msg, chatbot])
    demo.launch(share=False)
```
